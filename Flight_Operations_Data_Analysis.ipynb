#### Data Cleaning and Missing Value Handling

Before performing further analysis, missing values need to be handled to ensure that the dataset is suitable for analysis. The missing values in the `Delay_Minutes` column are replaced using the median value of that column.

```python
# Check missing values before handling
df.isnull().sum()
```

```python
# Fill missing values
df["Delay_Minutes"] = df["Delay_Minutes"].fillna(df["Delay_Minutes"].median())
```

After handling the missing values, the dataset is checked again to confirm that no missing values remain.

```python
df.isnull().sum()
```

#### Data Selection and Filtering

Data selection and filtering are used to focus on specific columns and records that are relevant to the analysis.

##### Selecting Relevant Columns

Relevant columns are selected to examine important information such as the airline, destination, number of passengers, flight delay, and passenger satisfaction.

```python
df[[
    "Airline",
    "Origin",
    "Destination",
    "Passengers",
    "Delay_Minutes",
    "Passenger_Satisfaction"
]].head()
```

##### Selecting a Single Column

A single column can be selected to examine the values of a particular variable. The following operation displays the number of passengers for the first few flights.

```python
df["Passengers"].head()
```

##### Filtering Delayed Flights

Flights with a delay greater than zero minutes are filtered to examine the records of delayed flights.

```python
delayed_flights = df[df["Delay_Minutes"] > 0]

delayed_flights.head()
```

##### Filtering Flights Based on Passenger Satisfaction

The passenger satisfaction column is examined to identify the available satisfaction categories before applying a filter.

```python
df["Passenger_Satisfaction"].unique()
```

##### Filtering Flights with More Than 200 Passengers

Flights carrying more than 200 passengers are filtered to focus on high-passenger flight operations.

```python
high_passenger_flights = df[df["Passengers"] > 200]

high_passenger_flights.head()
```

##### Filtering Using Multiple Conditions

Multiple conditions can be combined to select specific flight records. The following operation filters flights with more than 200 passengers and a delay of more than 30 minutes.

```python
filtered_flights = df[
    (df["Passengers"] > 200) &
    (df["Delay_Minutes"] > 30)
]

filtered_flights.head()
```

#### Sorting the Data

Sorting helps organize the dataset based on specific columns. In this section, the flight records are sorted based on delay duration and passenger count to identify notable patterns.

##### Sorting Flights by Delay Duration

The dataset is sorted in descending order of delay minutes to identify the flights with the highest delays.

##### Sorting Flights by Number of Passengers

The dataset is sorted in descending order based on the number of passengers to identify the flights carrying the highest number of passengers.

```python
df.sort_values(by="Passengers", ascending=False).head()
```

#### Grouping and Aggregation

Grouping and aggregation are used to analyze flight operations based on different categories. These operations help identify patterns by calculating summary values such as average delays, total passengers, and the number of flights.

##### Average Delay by Airline

The average delay for each airline is calculated to compare the operational performance of different airlines.

```python
average_delay_by_airline = df.groupby("Airline")["Delay_Minutes"].mean()

average_delay_by_airline.sort_values(ascending=False)
```

##### Total Passengers by Airline

The total number of passengers is calculated for each airline to identify which airline carried the highest number of passengers.

```python
total_passengers_by_airline = df.groupby("Airline")["Passengers"].sum()

total_passengers_by_airline.sort_values(ascending=False)
```

##### Number of Flights by Destination

The number of flight records for each destination is calculated to identify the most frequently occurring destinations in the dataset.

```python
flights_by_destination = df.groupby("Destination").size()

flights_by_destination.sort_values(ascending=False)
```

#### Data Transformation

Data transformation is used to create or modify columns so that the dataset can provide additional insights. In this section, new columns are created based on existing flight operation data.

##### Creating a Delay Category

A new column named `Delay_Category` is created to classify flights based on their delay duration.

```python
def categorize_delay(delay):
    if delay == 0:
        return "On Time"
    elif delay <= 30:
        return "Low Delay"
    else:
        return "High Delay"

df["Delay_Category"] = df["Delay_Minutes"].apply(categorize_delay)

df[["Delay_Minutes", "Delay_Category"]].head()
```

##### Creating a Passenger Category

A new column named `Passenger_Category` is created to classify flights based on the number of passengers.

```python
def categorize_passengers(passengers):
    if passengers < 150:
        return "Low"
    elif passengers <= 250:
        return "Medium"
    else:
        return "High"

df["Passenger_Category"] = df["Passengers"].apply(categorize_passengers)

df[["Passengers", "Passenger_Category"]].head()
```

#### Analysis

The analysis helps identify important patterns and trends in flight operations, such as airline delays, passenger traffic, and destination frequency.

##### Airline with the Highest Average Delay

The average delay is calculated for each airline to identify which airline has the highest average flight delay.

```python
average_delay_by_airline = df.groupby("Airline")["Delay_Minutes"].mean()

average_delay_by_airline.sort_values(ascending=False)
```

```python
average_delay_by_airline.sort_values(ascending=False).head(1)
```

*IndiGo has the highest average delay of approximately 25.03 minutes among the airlines in the dataset. This indicates that IndiGo experienced comparatively higher average flight delays during the period represented by the dataset.*

##### Airline with the Highest Number of Passengers

The total number of passengers is calculated for each airline to identify which airline handled the highest number of passengers in the dataset.

```python
total_passengers_by_airline = df.groupby("Airline")["Passengers"].sum()

total_passengers_by_airline.sort_values(ascending=False)
```

*Vistara has the highest total number of passengers with 4,484 passengers in the dataset. This indicates that Vistara handled the highest passenger volume among the airlines included in the dataset.*

##### Most Frequent Destination

The number of flights is calculated for each destination to identify which destination appears most frequently in the dataset.

```python
flights_by_destination = df.groupby("Destination").size()

flights_by_destination.sort_values(ascending=False)
```

```python
flights_by_destination.sort_values(ascending=False).head(1)
```

*Delhi is the most frequent destination in the dataset, with 51 flight records. This indicates that Delhi had the highest number of flights among the destinations included in the dataset.*

##### Distribution of Flights by Delay Category

The number of flights in each delay category is calculated to understand the overall distribution of on-time and delayed flights.

```python
delay_category_counts = df["Delay_Category"].value_counts()

delay_category_counts
```

*The dataset contains 70 on-time flights, 70 flights with low delays, and 40 flights with high delays. This shows that on-time and low-delay flights are equally common, while high-delay flights are less frequent.*

##### Distribution of Flights by Passenger Category

The number of flights in each passenger category is calculated to understand the distribution of flights based on passenger volume.

```python
passenger_category_counts = df["Passenger_Category"].value_counts()

passenger_category_counts
```

*The dataset contains 120 flights in the Low passenger category and 60 flights in the Medium passenger category. No flights fall into the High passenger category based on the defined classification. This indicates that most flights in the dataset carried fewer than 150 passengers.*

##### Average Passengers by Delay Category

The average number of passengers is calculated for each delay category to examine whether passenger volume varies with flight delay levels.

```python
average_passengers_by_delay = df.groupby("Delay_Category")["Passengers"].mean()

average_passengers_by_delay
```

*The average passenger count is highest for low-delay flights at approximately 123.90 passengers, followed by high-delay flights at approximately 123.33 passengers. On-time flights have the lowest average passenger count at approximately 121.33 passengers. The differences are relatively small, indicating that passenger volume does not vary significantly across the delay categories in this dataset.*
